# Week 8 — Tell the Story

## Five-minute demo outline

**0:00–0:35 — Question.** Content teams cannot inspect every page. Which pages should enter a first human discoverability review?

**0:35–1:25 — Data and safety.** Use public-safe aggregate Search Console fields from the FlyRank internship warehouse. Exclude client names, domains, URLs, credentials, private queries, identifiers as features, future inputs, and label-derived fields.

**1:25–2:20 — Method.** Apply the transparent baseline `impressions × (1 − CTR)` to pages with at least 100 impressions and average position from 5 to 20. Compare it with a compact regularized score on the same grouped, time-ordered held-out slice.

**2:20–3:15 — One chart.** Show the paper's precision@10 comparison: 0.60 for the rule baseline and 0.80 for the regularized score. Explain that the comparison is directional and depends on the review labels and split.

**3:15–4:05 — Honest result.** The learned score produced a modest held-out ranking lift while preserving an interpretable decision boundary. This does not prove that editing a page causes higher rankings or traffic.

**4:05–5:00 — Recommendation.** Review high-impression, mid-position, under-captured-click pages first; then check query intent, SERP features, volatility, and technical context before changing a title or description.
